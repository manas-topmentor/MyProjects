# 2.Find the number of users by gender.

```python
df1['gender'].value_counts()
```

```python
import matplotlib.pyplot as plt

# 1. Define the data
categories = df1['gender'].value_counts()
sizes = df1.groupby('gender').size() # The values for each slice

# 2. Configure styling options
colors = ['#ff9999', '#66b3ff', '#99ff99', '#ffcc99', '#c2c2f0']

# 3. Create the pie chart
plt.figure(figsize=(7, 7))  # Set the window/figure size
plt.pie(
    sizes, 
    labels=categories, 
    colors=colors, 
    shadow=True         # Add a slight drop shadow for depth
)

# 4. Final touches and display
plt.title('USER BY GENDER', fontsize=12, fontweight='bold')

plt.show()
```

```python
import matplotlib.pyplot as plt

categories = df1.devices_used.unique()
values = df1.devices_used.value_counts()
fig, ax = plt.subplots(figsize=(8, 5))
ax.barh(y=categories, width=values, color='skyblue', edgecolor='black')
ax.set_xlabel('USERS')
ax.set_ylabel('DEVICES')
ax.set_title('DEVICES VS USER')

# 5. Optional: Invert the y-axis so the first item appears at the top
ax.invert_yaxis()

# 6. Display the plot
plt.tight_layout()
plt.show()
```
