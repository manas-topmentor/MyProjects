```python
from datetime import datetime

# List to store books
books = []

# Dictionary to store issued books
issued_books = {}

# ---------------------- Functions ----------------------

def add_book():
    book_id = input("Enter Book ID: ")
    title = input("Enter Book Title: ")
    author = input("Enter Author Name: ")

    book = {
        "id": book_id,
        "title": title,
        "author": author,
        "available": True
    }

    books.append(book)
    print("Book added successfully!\n")


def view_books():
    if len(books) == 0:
        print("No books available.\n")
        return

    print("\n----- Library Books -----")
    for book in books:
        status = "Available" if book["available"] else "Issued"
        print(f"ID: {book['id']}")
        print(f"Title: {book['title']}")
        print(f"Author: {book['author']}")
        print(f"Status: {status}")
        print("-------------------------")
    print()


def issue_book():
    book_id = input("Enter Book ID to Issue: ")

    for book in books:
        if book["id"] == book_id:
            if book["available"]:
                student = input("Enter Student Name: ")
                issue_date = datetime.now()

                # Tuple (Book ID, Student Name, Issue Date)
                issued_books[book_id] = (book_id, student, issue_date)

                book["available"] = False

                print("Book issued successfully!\n")
                return
            else:
                print("Book is already issued.\n")
                return

    print("Book not found.\n")


def view_issued_books():
    if len(issued_books) == 0:
        print("No books issued.\n")
        return

    print("\n----- Issued Books -----")

    for book_id, details in issued_books.items():
        print(f"Book ID : {details[0]}")
        print(f"Student : {details[1]}")
        print(f"Issue Date : {details[2].strftime('%d-%m-%Y')}")
        print("------------------------")
    print()


def return_book():
    book_id = input("Enter Book ID to Return: ")

    if book_id not in issued_books:
        print("This book is not issued.\n")
        return

    return_date = datetime.now()

    issue_info = issued_books[book_id]
    issue_date = issue_info[2]

    days = (return_date - issue_date).days

    fine = 0
    if days > 7:
        fine = (days - 7) * 10

    for book in books:
        if book["id"] == book_id:
            book["available"] = True
            break

    del issued_books[book_id]

    print("Book returned successfully.")
    print("Days Kept:", days)

    if fine > 0:
        print("Late Fine = ₹", fine)
    else:
        print("No Fine.")
    print()


# ---------------------- Main Menu ----------------------

while True:
    print("===== LIBRARY MANAGEMENT SYSTEM =====")
    print("1. Add Book")
    print("2. View Books")
    print("3. Issue Book")
    print("4. Return Book")
    print("5. View Issued Books")
    print("6. Exit")

    choice = input("Enter your choice: ")

    if choice == "1":
        add_book()

    elif choice == "2":
        view_books()

    elif choice == "3":
        issue_book()

    elif choice == "4":
        return_book()

    elif choice == "5":
        view_issued_books()

    elif choice == "6":
        print("Thank you for using the Library Management System.")
        break

    else:
        print("Invalid Choice! Please try again.\n")
```
